## SECTION 6: Common Failures - Real Production Issues

### Failure 1: Rotation Race Condition

**How it breaks:**
```python
# Request A starts with old key
openai_client = OpenAI(api_key=fetch_from_vault())

# Rotation happens mid-request
vault.rotate_secret("openai_key", new_key)

# Request A completes - API call fails with 401

# Error: Incorrect API key provided: sk-old-***
```

**Root cause:** Rotation updated Vault but didn't wait for in-flight requests to complete

**The fix:** Request tracking with grace period
```python
with rotation_manager.track_request():
    response = openai_client.chat.completions.create(...)
```

**Production impact:** 10-20 requests failed with 401 errors during rotation


**Root cause:** Git tracks all history permanently. Removing file ≠ removing history

**The fix:**

## SECTION 5: Reality Check - What This DOESN'T Do

**CRITICAL:** Let's be honest about limitations. Vault is powerful, but it's not magic.

### What This DOESN'T Do

#### 1. Doesn't prevent all secret leaks

Vault prevents secrets in git and on disk, but **can't stop a developer from logging secrets** to console or CloudWatch.

**Example:**
```python
logger.debug(f"Using key: {api_key}")  # ❌ Bypasses Vault, logs the secret
```

**Workaround:** Implement log filtering that redacts patterns matching `sk-.*` or other API key formats.

**Reality:** Secret scanning catches 80% of leaks, not 100%.

#### 2. Doesn't eliminate key rotation complexity

We achieved zero downtime, but **rotation still requires coordinating with external services**.

**Problem:** If you rotate your OpenAI key in Vault but forget to update it in OpenAI's dashboard, requests fail immediately.

**Why:** Vault can't know about every SaaS provider's rotation API. You must manually:
1. Generate new key with OpenAI
2. Update in Vault

**Impact:** Get the order wrong → outage

#### 3. Doesn't scale infinitely

Our LRU cache holds 128 secrets in memory. At high scale:
- 10,000 requests/second with 500 unique secret paths
- Cache hit rate drops to 25%
- 7,500 Vault API calls per second
- **OSS Vault handles ~500 req/sec before degradation**

**When you'll hit this:** Multi-tenant SaaS with per-customer secrets or 100+ microservices

**What to do instead:** Use Vault Agent with sidecar pattern or upgrade to Vault Enterprise

- **Complexity:** Added 300+ lines of code, 2 new dependencies, and a Vault server to run/maintain
- **Performance:** Added 5-15ms latency on startup, 2-3ms on health checks

**Scenario:** 100+ microservices each needing 20+ secrets, or regulated environment requiring HSM-backed encryption

**What breaks:**
- OSS Vault becomes single point of failure
- Secret retrieval latency increases from 20ms to 200ms+
- Managing Vault upgrades, backups, DR becomes full-time job

## SECTION 3: Theory Foundation

### Core Concept: Enterprise Secrets Management

Think of secrets management like a **bank vault versus hiding cash under your mattress**:
- Your `.env` file is the mattress - convenient but risky
- HashiCorp Vault is the bank vault - secure, audited, and controlled access

### How Vault Works

**Step 1: Centralized Storage**  
All secrets live in one place with encryption at rest. Your application never stores secrets - it requests them at runtime from Vault.

**Step 2: Dynamic Access**  
Applications authenticate to Vault using identity (not stored credentials). Vault issues short-lived tokens specific to that application's needs.

**Step 3: Automatic Rotation**  
Vault can automatically rotate secrets and notify applications. Your OpenAI key rotates every 30 days without manual intervention.

**Step 4: Audit Trail**  
Every secret access is logged with identity, timestamp, and purpose. You can see exactly who accessed production database passwords and when.

### Why This Matters for Production

- **Security:** Secrets never touch disk or git. If an engineer's laptop is stolen, no secrets are compromised because they expired 1 hour after being fetched.
