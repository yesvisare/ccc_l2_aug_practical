## 8. Common Failure Modes

### 1. Feedback Bias ⚠️
**Problem:** Only unhappy users respond → Skewed dataset

**Example:**
- 1000 queries/day
- 80% users satisfied (don't respond)
- 20% users dissatisfied (all respond)
- Result: 100% negative feedback!

**Fix:**
- Proactive sampling of neutral queries
- Incentivize all feedback (not just negative)
- Compare to random baseline annotations

### 2. Low IAA ⚠️
**Problem:** Annotators disagree (IAA < 0.70) → Unreliable labels

**Causes:**
- Vague annotation guidelines
- Ambiguous edge cases
- Insufficient training

**Fix:**
- Add concrete examples to guidelines
- Calibration sessions with gold labels
- Simplify annotation task

### 3. Active Learning Selects Wrong Queries ⚠️
**Problem:** Annotators waste time on trivial/duplicate queries

**Fix:**
- Increase diversity clustering
- Add deduplication step
- Manual review of top-priority queue

### 4. Annotation Bottleneck ⚠️
**Problem:** Backlog grows faster than annotation capacity

**Fix:**
- Reduce daily target (50 → 30)
- Hire more annotators
- Automate pre-filtering

### 5. Loop Never Closes ✗
**Problem:** Annotations collected but system never improves

**Fix:**
- Schedule weekly retraining
- Set up automated alerts
- Create improvement dashboard

## 7. Step 5: Feedback Loop Closure

**Critical:** Collecting annotations is useless unless they improve your system!

### Three Closure Mechanisms:

**1. Aggregate Annotations**
   - Combine multiple annotator labels (majority vote)
   - Compute confidence based on agreement
   - Filter high-confidence examples (>0.7)

**2. Extract Training Examples**
   - Use high-confidence labels to retrain embeddings
   - Update prompt templates based on patterns
   - Add corrected examples to few-shot prompts

**3. Monitor Improvement**
   - Track metrics before/after retraining
   - Measure feedback response rate trend
   - Alert on IAA degradation

**Failure Mode:** Loop never closes → Wasted annotation budget

````python
# Initialize IAA calculator
iaa = InterAnnotatorAgreement()

# Extract annotations from example data (2 annotators on same queries)
annotations_df = pd.DataFrame(example_data['annotations'])

# Filter for query q_001 (has 2 annotations)
q001_anns = annotations_df[annotations_df['query_id'] == 'q_001']
annotator_alice = q001_anns[q001_anns['annotator_id'] == 'annotator_alice']['factual_correctness'].values
annotator_bob = q001_anns[q001_anns['annotator_id'] == 'annotator_bob']['factual_correctness'].values

# Similarly for q_002
q002_anns = annotations_df[annotations_df['query_id'] == 'q_002']
alice_q002 = q002_anns[q002_anns['annotator_id'] == 'annotator_alice']['factual_correctness'].values
bob_q002 = q002_anns[q002_anns['annotator_id'] == 'annotator_bob']['factual_correctness'].values

# Combine into full annotation vectors
alice_all = [int(annotator_alice[0]), int(alice_q002[0])]
bob_all = [int(annotator_bob[0]), int(bob_q002[0])]

# Calculate Cohen's Kappa
kappa = iaa.cohens_kappa(alice_all, bob_all)

print(f"✓ Inter-Annotator Agreement (Cohen's Kappa)")
print(f"  κ = {kappa:.3f}")

if kappa >= 0.70:
    print(f"  ✓ Good agreement (≥0.70)")
elif kappa >= 0.40:
    print(f"  ⚠ Moderate agreement (0.40-0.69)")
else:
    print(f"  ✗ Poor agreement (<0.40)")

# Calculate Krippendorff's Alpha for multiple annotators
alpha = iaa.krippendorffs_alpha([alice_all, bob_all])
print(f"\n✓ Krippendorff's Alpha: {alpha:.3f}")

# Expected: Kappa ≈ 1.0 (perfect agreement in example)
````

## 6. Step 4: Inter-Annotator Agreement (IAA)

**Problem:** How do we know if annotations are reliable?

**Solution:** Measure consistency between multiple annotators.

### Metrics:

**1. Cohen's Kappa (κ)**: For 2 annotators
- Formula: `κ = (observed_agreement - expected_agreement) / (1 - expected_agreement)`
- Interpretation:
  - `κ > 0.70`: Good agreement ✓
  - `0.40 < κ < 0.70`: Moderate agreement ⚠
  - `κ < 0.40`: Poor agreement ✗

**2. Krippendorff's Alpha (α)**: For multiple annotators
- More robust for missing data
- Similar interpretation to Kappa

**Action Thresholds:**
- IAA < 0.70 → Refine annotation guidelines
- IAA < 0.40 → Retrain annotators or simplify task

## 4. Step 2: Active Learning Prioritization

**Challenge:** You have 1,000 queries/day but can only annotate 50.

**Solution:** Prioritize high-value queries using:

1. **Uncertainty Sampling**: Select queries with low model confidence
   - `uncertainty_score = 1 - confidence`

2. **Negative Feedback Boosting**: Prioritize queries with thumbs-down
   - `priority *= (1 + boost_factor)`  (default: 50% boost)

3. **Diversity Clustering**: Use K-means to avoid redundant annotations
   - Select top queries from each cluster

**Cost Trade-off:** 
- More annotation = Better labels but higher cost ($0.50-$2/query)
- Fewer annotation = Lower cost but slower improvement
